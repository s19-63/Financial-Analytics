# Lab 4: Data Cleaning
# Missing Values

```python
df.isnull().sum()
df.dropna(inplace=True)
```

# Duplicate Values

```python
df.duplicated().sum()
df.drop_duplicates(inplace=True)
```

# Lab 5: Profit Calculation

```python
df['Profit'] = df['Revenue'] - df['Expenses']

df.head()
```

# Lab 6: Total Revenue Analysis

```python

total_revenue = df['Revenue'].sum()
print(total_revenue)
```

# Visualization

```python

plt.figure(figsize=(8,5))

plt.bar(
["Revenue"],
[total_revenue]
)

plt.title("Total Revenue")
plt.show()
```

# Lab 7: Total Expense Analysis

```python

total_expense = df['Expenses'].sum()

print(total_expense)
plt.figure(figsize=(8,5))

plt.bar(
["Expenses"],
[total_expense]
)

plt.title("Total Expenses")
plt.show()
```

# Lab 8: Total Profit Analysis

```python

total_profit = df['Profit'].sum()

print(total_profit)
plt.figure(figsize=(8,5))

plt.bar(
["Profit"],
[total_profit]
)

plt.title("Total Profit")
plt.show()
```

# Lab 9: Monthly Revenue Trend

```python

df['Date'] = pd.to_datetime(df['Date'])

monthly_revenue = df.groupby(df['Date'].dt.month)['Revenue'].sum()


monthly_revenue
```

# monthly_revenue
# Visualization


```python

monthly_revenue.plot(
kind='line',
marker='o',
figsize=(10,5)
)

plt.title("Monthly Revenue Trend")
plt.show()
```

# Business Insight
•	Best Revenue Month? 

•	Lowest Revenue Month? 

# Lab 10: Monthly Expense Trend

```python
monthly_expense = df.groupby(
df['Date'].dt.month
)['Expenses'].sum()
monthly_expense.plot(
kind='line',
marker='o'
)

plt.title("Monthly Expense Trend")
plt.show()
```

# Lab 11: Monthly Profit Trend

```python
monthly_profit = df.groupby(
df['Date'].dt.month
)['Profit'].sum()
monthly_profit.plot(
kind='line',
marker='o'
)

plt.title("Monthly Profit Trend")
plt.show()
```

# Lab 12: Department Wise Revenue

```python
dept_revenue = df.groupby(
'Department'
)['Revenue'].sum()

dept_revenue
dept_revenue.plot(
kind='bar',
figsize=(10,5)
)

plt.title("Department Revenue")
plt.show()
```

# Lab 13: Department Wise Profit

```python
dept_profit = df.groupby(
'Department'
)['Profit'].sum()

dept_profit
dept_profit.plot(
kind='bar',
figsize=(10,5)
)

plt.title("Department Profit")
plt.show()
```

# Business Insight

•	Lowest Profit Department? 


# Lab 14: Budget vs Actual Analysis

```python
budget_actual = df.groupby(
'Department'
)[['Budget','Expenses']].sum()

budget_actual
```

# Visualization

```python
budget_actual.plot(
kind='bar',
figsize=(10,5)
)

plt.title("Budget vs Actual")
plt.show()
```

# Business Insight
•	Which departments exceeded budget? 

•	Which departments stayed within budget? 


# Lab 15: Profit Margin Analysis

```python
df['Profit_Margin'] = (
df['Profit'] /
df['Revenue']
) * 100

df['Profit_Margin'].mean()
```

# Business Insight
•	Average Profit Margin? 

•	Healthy or Poor Margin? 


# Lab 16: Financial KPI Dashboard

```python
total_revenue = df['Revenue'].sum()

total_expenses = df['Expenses'].sum()

total_profit = df['Profit'].sum()

profit_margin = (
total_profit /
total_revenue
)*100
print("Revenue:",total_revenue)
print("Expenses:",total_expenses)
print("Profit:",total_profit)
print("Profit Margin:",profit_margin)
```

# Lab 17: Correlation Analysis

```python
corr = df[
['Revenue',
'Expenses',
'Profit',
'Budget']
].corr()
sns.heatmap(
corr,
annot=True,
cmap='coolwarm'
)

plt.show()
```

# Insight
•	Revenue & Profit Relationship? 

•	Budget & Expense Relationship? 


# Lab 18: Yearly Financial Trend

```python
yearly = df.groupby(
df['Date'].dt.year
)[['Revenue',
'Expenses',
'Profit']].sum()

yearly
yearly.plot(
kind='bar',
figsize=(12,5)
)

plt.title("Yearly Financial Performance")
plt.show()
```
